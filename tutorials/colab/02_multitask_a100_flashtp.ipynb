# SevenNet-Polar · Multitask E/F/S/BEC training and FlashTP

Train a small equivariant model on **real Li₃PO₄ DFT/DFPT data**, reload its checkpoint, and examine convergence and held-out parity. This tutorial adds energy, forces, stress, and optional accelerator validation.

**Colab:** choose Runtime → Change runtime type → A100 if available, then Runtime → Run all. A different CUDA GPU or CPU can run the standard path; A100 allocation is not guaranteed by a subscription. Local Jupyter needs PyTorch installed first. The notebook preserves that installation.

The compact model is an educational example, not a production potential. No dielectric model is trained. The [tutorial guide and validation records](https://github.com/AugustinLu/SevenNet-Polar/tree/main/tutorials/colab) explain current test coverage.

## 1. Install the Polar fork and inspect the runtime

The package import is called `sevenn`, as in upstream SevenNet. Setup installs this repository and verifies its origin. A fresh Colab session already includes PyTorch. Installation output is saved to a file; a failure prints its final lines. All runs use float32, including T4 (no BF16 assumptions).

````python
import importlib.metadata
import json
import os
from pathlib import Path
import subprocess
import sys
import time
import uuid

START = time.perf_counter()
REPOSITORY = 'https://github.com/AugustinLu/SevenNet-Polar.git'
REPO_REF = os.environ.get(
    'SEVENNET_TUTORIAL_REF', 'main'
)  # report.json records the exact source commit.

# Local Jupyter reuses a checkout. Colab fetches the public main branch.
previous_repo = globals().get('REPO')
local = os.environ.get('SEVENNET_TUTORIAL_REPO')
candidates = (
    ([Path(local)] if local else [])
    + ([previous_repo] if previous_repo else [])
    + list(Path.cwd().resolve().parents)
    + [Path.cwd(), Path.cwd() / 'SevenNet-Polar']
)
REPO = next(
    (
        p.resolve()
        for p in candidates
        if (p / 'tutorials/colab/support.py').is_file()
    ),
    None,
)
fetch_started = time.perf_counter()
if REPO is None:
    REPO = Path.cwd() / ('SevenNet-Polar-' + uuid.uuid4().hex[:8])
    subprocess.run(
        [
            'git',
            'clone',
            '--depth',
            '1',
            '--branch',
            REPO_REF,
            REPOSITORY,
            str(REPO),
        ],
        check=True,
    )
FETCH_SECONDS = time.perf_counter() - fetch_started


TORCH_BEFORE = importlib.metadata.version('torch')
WORK = Path.cwd() / ('polar-tutorial-' + uuid.uuid4().hex[:8])
WORK.mkdir()
INSTALL_LOG = WORK / 'installation.log'
# Constrain PyTorch to the runtime wheel. Pip retains already-satisfied dependencies.
constraint = WORK / 'runtime-constraint.txt'
constraint.write_text('torch==' + TORCH_BEFORE + '\n')
install_started = time.perf_counter()
SKIP_INSTALL = (
    os.environ.get('SEVENNET_TUTORIAL_SKIP_INSTALL') == '1'
)  # Test harness only.
if not SKIP_INSTALL:
    with INSTALL_LOG.open('w') as log:
        result = subprocess.run(
            [
                sys.executable,
                '-m',
                'pip',
                'install',
                '--quiet',
                '--constraint',
                str(constraint),
                '--editable',
                str(REPO),
                'matplotlib',
            ],
            stdout=log,
            stderr=subprocess.STDOUT,
        )
    if result.returncode:
        print(INSTALL_LOG.read_text()[-5000:])
        raise RuntimeError(f'Installation failed. Full log: {INSTALL_LOG}')
INSTALL_SECONDS = time.perf_counter() - install_started
if importlib.metadata.version('torch') != TORCH_BEFORE:
    raise RuntimeError('PyTorch changed. Restart the runtime before continuing.')
if (
    'sevenn' in sys.modules
    and REPO not in Path(sys.modules['sevenn'].__file__).resolve().parents
):
    raise RuntimeError(
        'Another sevenn was already imported. '
        'Restart the kernel and run this notebook first.'
    )
sys.path.insert(0, str(REPO))
sys.path.insert(0, str(REPO / 'tutorials/colab'))
import support
from IPython.display import display

ENVIRONMENT = support.environment(REPO)
DEVICE = ENVIRONMENT['device']
print(json.dumps(ENVIRONMENT, indent=2))
print(
    f'Fetch: {FETCH_SECONDS:.1f} s · install: {INSTALL_SECONDS:.1f} s '
    '· PyTorch retained'
)
print(f'Artifacts: {WORK}')
if DEVICE == 'cpu':
    print(
        'CPU selected. For Colab GPU: Runtime → Change runtime type → '
        'T4 or A100, then Run all.'
    )
````

## 2. Real labels and a fixed whole-structure split

The default uses **50 training, 20 validation and 20 test structures** from the author's official [SevenNet-Polar datasets](https://doi.org/10.5281/zenodo.21322761), described in [Lu et al. (2026)](https://arxiv.org/abs/2607.14827). A bounded 240-frame pool is bundled with source checksums, original frame IDs and [attribution](https://github.com/AugustinLu/SevenNet-Polar/blob/main/example_inputs/tutorial_polar/DATA.md). Training subsets are nested; validation and test structures stay fixed when the training count changes. Selection balances compositions without using label values or model errors.

Li₃PO₄ includes stoichiometric and vacancy compositions. Li₁₁O₁₆P₄ reference BEC tensors sum to approximately **1 e per cell**. A charged-cell DFPT convention is the likely explanation, but the original settings have not been independently verified. We preserve these labels. Inspect both raw tensor sums and prediction-minus-reference tensor-sum errors; a zero target is inappropriate for those frames.

All E/F/S/BEC labels survive serialization checks, and geometry fingerprints screen duplicates. Trajectory/parent IDs are unavailable, so the held-out errors demonstrate the workflow rather than independent scientific generalization. Splitting keeps whole structures together; no test labels enter training statistics or checkpoint selection. Dielectric labels are excluded.

Set `DATASET='zro2'` for the alternative ZrO₂ pool. Its reference sums are close to zero, but the broader small-subset fits showed larger prediction residuals. Increasing the training count up to 200 adds data and runtime; it does not guarantee a smaller worst-case residual.


````python
DATASET = os.environ.get('SEVENNET_TUTORIAL_DATASET', 'li3po4')
TRAINING_STRUCTURES = int(
    os.environ.get('SEVENNET_TUTORIAL_TRAINING_STRUCTURES', '50')
)
DATA_DIR = WORK / 'data'
structures, split_paths, all_path, split_manifest, split_summary = (
    support.prepare_subset_data(REPO, DATA_DIR, TRAINING_STRUCTURES, dataset=DATASET)
)
display(split_summary)
print(
    f'{len(structures)} structures; '
    f'{min(map(len, structures))}–{max(map(len, structures))} atoms per cell'
)
print('Pool SHA256:', split_manifest['source_sha256'])
print('BEC and E/F/S label round trips passed. Duplicate geometry screen passed.')
````

## 3. Configure the model

`lmax=2` retains tensor information; four message-passing layers, batch size 4 and 100 epochs provide a small training exercise. The multitask model fits E/F/S/BEC together with 32 channels on CUDA and 16 on CPU. FlashTP requires channels divisible by 32; the CPU fallback uses standard e3nn. Species reference energies are estimated from training data by the existing CLI option.

The best checkpoint minimizes validation weighted total loss, using the package's established loss definitions. ASR is a diagnostic and does not affect the objective or checkpoint selection. With batch size 4, 50 structures give 13 optimizer updates per epoch. Reducing epochs is useful for a plumbing check but changes the fitting budget. CPU users may start with 1–5 epochs and then select T4 for the complete fit.


````python
# Adjust these few controls; keep the test split fixed.
EPOCHS = int(os.environ.get('SEVENNET_TUTORIAL_EPOCHS', '100'))
BATCH_SIZE = 4
CHANNELS = 32 if DEVICE == 'cuda' else 16  # FlashTP needs multiples of 32.
LAYERS = 4
SEED = 42
MULTITASK = True
CONFIG = support.make_config(
    split_paths,
    EPOCHS,
    multitask=MULTITASK,
    batch_size=BATCH_SIZE,
    channel=CHANNELS,
    layers=LAYERS,
    device=DEVICE,
    seed=SEED,
)
import yaml

print(yaml.safe_dump(CONFIG, sort_keys=False))
````

## 4. Train and inspect convergence

Training uses `sevenn train` through a profiling wrapper. Progress appears about every 30 seconds; full logs and epoch CSVs are retained. Wall time includes preprocessing, model construction and checkpoint writing. GPU memory is the process's peak PyTorch allocation/reservation, not total driver memory. A missing GPU uses CPU.

Force/stress convergence plots use component RMSE, matching the parity tables. The raw CLI CSV uses vector/tensor norm RMSE: divide force by √3 and stress by 10√6 to obtain component RMSE in eV/Å and GPa. Training curves accumulate predictions while weights change; parity uses the fixed best checkpoint.

The convergence figure also reports **raw ASR maximum and mean over structures**, in e per cell. These diagnostics do not change the loss or project the predictions. Validation uses one fixed model state per epoch; training aggregates successive optimizer states.

Dotted lines show the validation-label tensor-sum baseline. A nonzero baseline belongs to the supplied labels and is retained; predictions are never shifted to hide a residual.


````python
RUN = WORK / 'training'
CHECKPOINT, TRAIN_PROFILE = support.train(REPO, RUN, CONFIG)
print('Best validation checkpoint:', CHECKPOINT.name)
print(json.dumps(TRAIN_PROFILE, indent=2))
````

````python
curve, convergence_figure = support.learning_curves(
    RUN, multitask=MULTITASK, structures=structures, manifest=split_manifest
)
display(curve.tail())
display(convergence_figure)
import matplotlib.pyplot as plt

plt.close(convergence_figure)
````

## 5. Reload and evaluate the chosen checkpoint

The inference CLI reloads the **best validation checkpoint** and predicts the original frame order. Training, validation, and held-out test points have different colors. Tables report component-weighted RMSE and MAE; tensor components from one atom are not independent samples.

Diagonal and all six off-diagonal BEC entries are reported separately; BEC tensors need not be symmetric. Energy is reported per atom (eV/atom), forces in eV/Å, and stress in GPa with ASE's tensile-positive sign. The CLI stores stress in kbar with the opposite sign; the plot converts both reference and prediction consistently.

````python
atom_predictions, graph_predictions, INFERENCE_PROFILE = support.evaluate(
    REPO, RUN, CHECKPOINT, all_path, DEVICE, batch_size=BATCH_SIZE
)
metrics, parity_figure = support.metrics_and_parity(
    atom_predictions, graph_predictions, split_manifest, RUN, multitask=MULTITASK
)
display(parity_figure)
plt.close(parity_figure)
display(metrics.round(5))
FIRST_PARITY_SECONDS = time.perf_counter() - START
print(
    f'Setup to first parity: {FIRST_PARITY_SECONDS:.1f} s '
    '(runtime allocation excluded)'
)
````

## 6. Use the checkpoint from Python and inspect the sum rule

`SevenNetCalculator` reloads the checkpoint for a test structure and is checked against the CLI. For each structure, the diagnostic is the maximum absolute component of **Σᵢ Z*ᵢ**, in e. The partition table reports the largest and median structure residuals; it does not sum different structures together or divide by atom count. These are raw predictions and raw DFT labels: no acoustic-sum-rule correction is applied. A small BEC RMSE alone does not guarantee a small cell-wide residual.

The helper checks this native Python usage against the inference CSV:

```python
from sevenn.calculator import SevenNetCalculator
atoms.calc = SevenNetCalculator(str(CHECKPOINT), device=DEVICE)
bec = atoms.calc.get_property('born_effective_charges', atoms)
```

The per-frame `asr.csv` also records atom counts, atom-normalized sums, and prediction-minus-reference tensor sums. Inspect the reference sums when using another dataset: a nonzero reference sum must not be mistaken for model error.

`sum_error_max_abs_e` is computed by subtracting each reference tensor sum from its prediction before taking maxima. Subtracting the two partition maxima would not give this error.

For this Li₃PO₄ pool, a raw maximum near 1 e can reflect the vacancy reference labels. Judge the model using the tensor-sum error and inspect composition-specific rows in `asr.csv`; do not interpret every nonzero sum as a neutral-cell ASR violation.


````python
structure_id = split_manifest['indices']['Test'][0]
predicted_bec = support.reload_ase(
    CHECKPOINT, structures[structure_id], DEVICE, atom_predictions, structure_id
)
print('ASE checkpoint reload agrees with CLI predictions.')
print(f"First atom's Z* in test structure {structure_id} (e):\n{predicted_bec[0]}")
asr = support.asr_diagnostics(
    atom_predictions, split_manifest, RUN, structures=structures
)
display(
    asr.groupby('partition').agg(
        reference_max_abs_sum_e=('reference_max_abs_sum_e', 'max'),
        prediction_max_abs_sum_e=('prediction_max_abs_sum_e', 'max'),
        prediction_median_abs_sum_e=('prediction_max_abs_sum_e', 'median'),
        sum_error_max_abs_e=('sum_error_max_abs_e', 'max'),
    )
)
````

## 7. Optional FlashTP: build, check, and measure

Set `TRY_FLASHTP=False` to skip the build. A CUDA GPU and compiler are required. The source revision is pinned, the architecture is detected, and build parallelism is limited to two jobs. CUDA compilation can dominate a short tutorial; it is timed separately. T4 support is established only if the checks below pass on a T4, not merely because the build accepts `sm_75`.

First compare **the same checkpoint and training batch** through e3nn and FlashTP: E/F/S/BEC outputs, weighted loss, and every parameter gradient, including the second derivatives needed for force/stress training. Then time warmed-up forward and forward/backward passes. Lastly run the same number of full training epochs from the same seed on both backends. The short learning curves and timings are exported. A failure is recorded explicitly and leaves the earlier standard model available.

A repeat in the same runtime reuses an installed build only when its clean source checkout matches the pinned revision. Reuse is labeled explicitly and has zero new build time; output/gradient checks still run. A fresh runtime builds from source.

````python
TRY_FLASHTP = os.environ.get('SEVENNET_TUTORIAL_FLASH', 'auto') != 'off'
FLASH_REVISION = 'd52e41454afc9636cd51757b65df44e7bcdca178'
COMPARISON_EPOCHS = 5
if not TRY_FLASHTP or DEVICE != 'cuda':
    FLASH_REPORT = {
        'status': 'skipped',
        'reason': 'disabled or CUDA GPU unavailable',
    }
    print(FLASH_REPORT)
else:
    from flash import install_flash, compare_backends

    FLASH_REPORT = {'status': 'started'}
    try:
        FLASH_REPORT['installation'] = install_flash(
            WORK / 'flashtp', FLASH_REVISION
        )
        FLASH_REPORT['comparison'] = compare_backends(
            CHECKPOINT, split_paths['Train'], batch_size=BATCH_SIZE
        )
        display(
            __import__('pandas').DataFrame(
                FLASH_REPORT['comparison']['measurements']
            )
        )
        equal_work = support.make_config(
            split_paths,
            COMPARISON_EPOCHS,
            multitask=True,
            batch_size=BATCH_SIZE,
            channel=CHANNELS,
            layers=LAYERS,
            device=DEVICE,
            seed=SEED,
        )
        FLASH_REPORT['training'] = {}
        for name, enabled in [('e3nn', False), ('FlashTP', True)]:
            cp, profile = support.train(
                REPO, WORK / ('compare_' + name), equal_work, flash=enabled
            )
            FLASH_REPORT['training'][name] = profile
        FLASH_REPORT['status'] = 'passed'
        base_time = FLASH_REPORT['training']['e3nn']['subprocess_wall_seconds']
        flash_time = FLASH_REPORT['training']['FlashTP']['subprocess_wall_seconds']
        print(
            f'Equal-work end-to-end ratio e3nn/FlashTP: '
            f'{base_time / flash_time:.2f} (build excluded)'
        )
    except Exception as error:
        FLASH_REPORT.update(
            status='failed', error_type=type(error).__name__, error=repr(error)
        )
        print('FlashTP validation failed; standard results remain available.')
        print(error)
    (RUN / 'flash_report.json').write_text(json.dumps(FLASH_REPORT, indent=2))
    # Include build logs and comparison curves; omit the large build caches.
    import shutil

    if (WORK / 'flashtp/build.log').exists():
        shutil.copy(WORK / 'flashtp/build.log', RUN / 'flash_build.log')
    for name in ['e3nn', 'FlashTP']:
        if (WORK / ('compare_' + name)).exists():
            for file in [
                'lc.csv',
                'training.log',
                'input.yaml',
                'training_profile.json',
            ]:
                path = WORK / ('compare_' + name) / file
                if path.exists():
                    shutil.copy(path, RUN / (name + '_' + file))
````
